```python
import pandas as pd

# Load the dataset with latin1 encoding to handle special characters
try:
    df = pd.read_csv('APL_Logistics.csv', encoding='latin1')
    print('Dataset loaded successfully!')
    display(df.head())
except Exception as e:
    print(f'Error: {e}')
```

```python
import numpy as np

# Identifying the product name column (usually 'Product Name')
product_col = [col for col in df.columns if 'Product Name' in col]

if product_col:
    col_name = product_col[0]
    unique_products = df[col_name].unique()
    print(f"Found {len(unique_products)} unique products in column: '{col_name}'")
    print("\nFirst 20 product names:")
    display(unique_products[:20])
else:
    print("Could not find a 'Product Name' column. Here are all available columns:")
    print(df.columns.tolist())
```

```python
# List the columns you want to remove here
columns_to_remove = ['Customer Fname', 'Customer Lname', 'Customer Street', 'Latitude', 'Longitude', 'Category Id', 'Department Id', 'Customer Id', 'Order Customer Id', 'Customer Zipcode', 'Product Name'] # Example columns

# Drop the columns. errors='ignore' prevents an error if the column was already deleted
df = df.drop(columns=columns_to_remove, errors='ignore')

print(f"Removed columns: {columns_to_remove}")
print(f"Remaining columns: {len(df.columns)}")
display(df.head())
```

```python
# Check for missing values
missing_data = df.isnull().sum()
if missing_data.any():
    print('Missing values found per column:')
    display(missing_data[missing_data > 0])
else:
    print('No missing values detected.')
```

```python
# Check for duplicate rows
duplicate_count = df.duplicated().sum()

if duplicate_count > 0:
    print(f'Found {duplicate_count} duplicate rows.')
    print('\nExample of duplicate rows:')
    display(df[df.duplicated()].head())
else:
    print('No duplicate rows found.')
```

```python
# Remove duplicate rows
initial_count = len(df)
df = df.drop_duplicates()
final_count = len(df)

print(f"Rows before: {initial_count}")
print(f"Rows after: {final_count}")
print(f"Total rows removed: {initial_count - final_count}")
```

```python
import pandas as pd

# Identify columns with string (object) data types
string_columns = df.select_dtypes(include=['object']).columns

for col in string_columns:
    # 1. Convert to string to avoid errors
    # 2. .strip() removes leading and trailing spaces
    # 3. .replace(r'\s+', ' ', regex=True) replaces multiple spaces with a single space
    df[col] = df[col].astype(str).str.strip().str.replace(r'\s+', ' ', regex=True)

print('Extra white spaces have been trimmed and normalized.')
# Verifying with the 'Order Region' column mentioned earlier
print('\nUnique values in Order Region:')
print(df['Order Region'].unique())
```

## Now we will do some changes in cleaned file

```python
import pandas as pd

df = pd.read_csv('cleaned_APL_Logistics.csv')
display(df.head())
```

```python
df['Delay_Gap'] = (
    df['Days for shipping (real)']
    - df['Days for shipment (scheduled)']
)
```

```python
def classify(delay):
    if delay < 0:
        return 'Early'
    elif delay == 0:
        return 'On Time'
    else:
        return 'Delayed'

df['Delivery_Class'] = df['Delay_Gap'].apply(classify)
```

```python
df['Delivery_Class'].value_counts()
```

```python
def severity(x):
    if x <= 0:
        return 'No Delay'
    elif x <= 2:
        return 'Low'
    elif x <= 5:
        return 'Medium'
    else:
        return 'High'

df['Delay_Severity'] = df['Delay_Gap'].apply(severity)
```

```python
# Calculate negative counts for 'Days for shipping (real)'
negative_real_shipping_days = (df['Days for shipping (real)'] < 0).sum()
print(f"Negative count for 'Days for shipping (real)': {negative_real_shipping_days}")
```

```python
# Calculate negative counts for 'Days for shipment (scheduled)'
negative_scheduled_shipping_days = (df['Days for shipment (scheduled)'] < 0).sum()
print(f"Negative count for 'Days for shipment (scheduled)': {negative_scheduled_shipping_days}")
```
