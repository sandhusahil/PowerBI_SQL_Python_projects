```python

# Step 2: Inspect dates and duplicate records

# Convert date column to datetime
df['date'] = pd.to_datetime(df['date'], dayfirst=True, errors='coerce')

print("Date data type:", df['date'].dtype)
print("Invalid dates:", df['date'].isna().sum())

print("\nDate range:")
print("Earliest date:", df['date'].min())
print("Latest date:", df['date'].max())

print("\nNumber of unique dates:", df['date'].nunique())

print("\nExact duplicate rows:", int(df.duplicated().sum()))

# Display duplicate records for inspection
display(df[df.duplicated(keep=False)].sort_values(
    by=['date', 'position']
).head(20))

# Inspect the missing song record
print("\nRecord with missing song:")
display(df[df['song'].isna()])
```

```python
display(df[df['song'].isna()].T)
```

```python

# Inspect all records for the same date
display(
    df[df['date'] == '2025-02-02']
    .sort_values('position')
    .head(5)
)

# Inspect records for the same artist
display(
    df[df['artist'].str.strip().str.lower() == 'lavern']
    .sort_values(['date', 'position'])
    .head(10)
)
```

```python
df = df.dropna(subset=['song'])
```

```python

# Check exact duplicate rows

print("Current dataset shape:", df.shape)
print("Exact duplicate rows:", int(df.duplicated().sum()))

# Display duplicate records for inspection
display(
    df[df.duplicated(keep=False)]
    .sort_values(by=['date', 'position'])
)
```

```python
df = df.drop_duplicates()
```

```python
# Remove the inconsistent date snapshot
df = df[df['date'] != pd.Timestamp('2025-03-01')]

print("Dataset shape:", df.shape)
print("Rows on 1 March 2025:", (df['date'] == pd.Timestamp('2025-03-01')).sum())
```

```python

# Final data-quality checks

print("Dataset shape:", df.shape)

print("\nMissing values:")
print(df.isna().sum())

print("\nExact duplicate rows:", int(df.duplicated().sum()))

print("\nUnique dates:", df['date'].nunique())

print("\nRows per date (summary):")
print(df.groupby('date').size().describe())

print("\nRows where position is outside 1–50:",
      (~df['position'].between(1, 50)).sum())

print("\nZero or negative duration:",
      (df['duration_ms'] <= 0).sum())

print("\nPopularity outside 0–100:",
      (~df['popularity'].between(0, 100)).sum())

print("\nAlbum types:")
print(df['album_type'].value_counts(dropna=False))
```

```python
daily_counts = df.groupby('date').size()
```

```python
display(daily_counts[daily_counts != 50].rename("row_count").to_frame())
```

```python
# Feature engineering

# Convert duration from milliseconds to minutes
df['duration_minutes'] = (df['duration_ms'] / 60000).round(2)

# Group playlist rankings
df['rank_band'] = pd.cut(
    df['position'],
    bins=[0, 10, 20, 30, 40, 50],
    labels=[
        'Top 10',
        '11–20',
        '21–30',
        '31–40',
        '41–50'
    ]
)

print("Updated dataset shape:", df.shape)

display(
    df[['date', 'position', 'song', 'duration_minutes', 'rank_band']].head()
)
```

```python
# cleaned dataset file

df.to_csv('Atlantic_France_Cleaned.csv', index=False)
```
